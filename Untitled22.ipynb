```python
"""
Compile a unified Indian-exam MCQ benchmark from HF datasets (run in Colab).
    pip install -q datasets pandas
Output: benchmark_base.jsonl  (input for the perturbation step)

Unified schema:
id, exam, year, subject, qtype, question, options(list|None), answer, has_numbers, source
"""
import re
import pandas as pd
from datasets import load_dataset

SEED = 42
N_PER_EXAM = {"NEET-PG": 300, "JEE-Adv": 250, "JEE-Main": 250}
SCHEMA = ["id", "exam", "year", "subject", "qtype", "question",
          "options", "answer", "has_numbers", "source"]
SUBJECT_MAP = {"Math": "Mathematics", "Chem": "Chemistry", "Phy": "Physics"}
NUM_RE = re.compile(r"\d+(\.\d+)?")

def has_numbers(text):
    return len(NUM_RE.findall(str(text))) >= 2

def stratified(df, n, by="subject"):
    if len(df) <= n:
        return df
    k = max(1, n // df[by].nunique())
    parts = [g.sample(min(len(g), k), random_state=SEED) for _, g in df.groupby(by)]
    return pd.concat(parts).sample(frac=1, random_state=SEED).head(n)

def to_letter(x):
    """normalize '1'/'1.0'/'A'/'a' -> 'A'... ; returns None if not a single option"""
    s = str(x).strip().upper()
    if s in ("A", "B", "C", "D"):
        return s
    try:
        i = int(float(s))
        if 1 <= i <= 4:
            return "ABCD"[i - 1]
    except ValueError:
        pass
    return None

# ---------- 1. NEET-PG : openlifescienceai/medmcqa ----------
def load_medmcqa():
    df = load_dataset("openlifescienceai/medmcqa", split="validation").to_pandas()
    df = df[(df["choice_type"] == "single") & (df["cop"].between(0, 3))]
    df = df[df["question"].str.len() > 20]
    return pd.DataFrame({
        "id": "medmcqa_" + df["id"].astype(str),
        "exam": "NEET-PG", "year": None,
        "subject": df["subject_name"], "qtype": "MCQ_SINGLE",
        "question": df["question"].str.strip(),
        "options": df[["opa", "opb", "opc", "opd"]].values.tolist(),
        "answer": df["cop"].map(lambda i: "ABCD"[int(i)]),
        "source": "openlifescienceai/medmcqa",
    })

# ---------- 2. JEE Advanced : daman1209arora/jeebench (2016-2023) ----------
def load_jeebench():
    df = load_dataset("daman1209arora/jeebench", split="test").to_pandas()
    yr = df["description"].str.extract(r"(20\d\d)")[0].astype(float)
    return pd.DataFrame({
        "id": "jeebench_" + df.index.astype(str),      # NOT df["index"]: it repeats across subjects
        "exam": "JEE-Adv", "year": yr,
        "subject": df["subject"].str.capitalize(),
        "qtype": df["type"],
        "question": df["question"].str.strip(),
        "options": None,                       # options embedded in question text
        "answer": df["gold"].astype(str),
        "source": "daman1209arora/jeebench",
    })

# ---------- 3. JEE Main : eQOURSE/jee-main-questions (3 configs, noisy) ----------
def load_eqourse():
    frames = []
    for cfg in ["chemistry", "physics", "mathematics"]:
        dd = load_dataset("eQOURSE/jee-main-questions", cfg)
        d = dd[list(dd.keys())[0]].to_pandas()
        d["_subject"] = cfg.capitalize()
        frames.append(d)
    df = pd.concat(frames, ignore_index=True)

    # drop image-dependent / garbled / too short-or-long rows
    q = df["question"].astype(str)
    bad = q.str.contains(r"\[IMAGE\]|¼|½", regex=True) | ~q.str.len().between(30, 1200)
    bad |= df["has_image"].astype(str).str.lower().isin(["true", "1"])
    df = df[~bad].copy()

    # MCQ rows -> options + letter answer ; numerical rows -> numeric answer, no options
    is_num = df["question_type"].astype(str).str.contains("numer", case=False)
    opts = df[["option_1", "option_2", "option_3", "option_4"]].astype(str).values.tolist()
    df["_opts"] = [None if n else o for n, o in zip(is_num, opts)]
    df["_ans"] = [
        str(na).strip() if n else to_letter(co)
        for n, na, co in zip(is_num, df["numerical_answer"], df["correct_option"])
    ]
    df = df[df["_ans"].notna() & ~df["_ans"].astype(str).str.lower().isin(["nan", "none", ""])]

    print("eQOURSE sample:", df[["question_type", "_ans"]].head(5).to_dict("records"))
    return pd.DataFrame({
        "id": "eq_" + df["question_id"].astype(str),
        "exam": "JEE-Main", "year": None,
        "subject": df["_subject"],
        "qtype": df["question_type"],
        "question": df["question"].astype(str).str.strip(),
        "options": df["_opts"],
        "answer": df["_ans"],
        "source": "eQOURSE/jee-main-questions",
    })

if __name__ == "__main__":
    parts = []
    for name, fn in [("NEET-PG", load_medmcqa), ("JEE-Adv", load_jeebench), ("JEE-Main", load_eqourse)]:
        try:
            df = fn()
        except Exception as e:
            print(f"[skip] {name}: {e}")
            continue
        if df is None or len(df) == 0:
            print(f"[skip] {name}: no rows")
            continue
        df["subject"] = df["subject"].replace(SUBJECT_MAP)
        df = df[df["subject"].notna() & (df["subject"] != "Unknown")]
        df = df.drop_duplicates("question")
        df["has_numbers"] = df["question"].map(has_numbers)
        df = stratified(df, N_PER_EXAM[name])
        print(name, len(df), df["subject"].value_counts().to_dict())
        parts.append(df.dropna(axis=1, how="all"))   # avoids concat FutureWarning

    bench = pd.concat(parts, ignore_index=True).reindex(columns=SCHEMA)
    assert bench["id"].is_unique, "duplicate ids in benchmark"
    bench.to_json("benchmark_base.jsonl", orient="records", lines=True, force_ascii=False)
    print("saved benchmark_base.jsonl", len(bench))
    print(bench.groupby("exam")["has_numbers"].agg(["count", "sum"]))
    print(bench.groupby(["exam", "qtype"]).size())
```

```python
"""
Step 2: label-preserving perturbations. Run in Colab after compile_benchmark.py.
perm[k] = index (in the original) of the option now shown at position k  -> lets eval check "same option chosen".
Input : benchmark_base.jsonl
Output: perturbed_twins.jsonl   (one row per (original question, perturbation))

Perturbations (gold answer stays valid -> safe for robustness / sensitivity):
  option_shuffle : permute options, remap gold letter(s)   [MCQ only]
  prefix_noise   : neutral instruction sentence prepended  [all rows]
Rows where option position matters ("all of the above", "both A and B" ...) are skipped for shuffle.
"""
import json, re, random
import pandas as pd

SEED = 42
rng = random.Random(SEED)

POSITIONAL = re.compile(
    r"(of the above|of these|both|neither|none of|all of"
    r"|(?:^|\s)\(?[a-d1-4]\)?\s*(?:and|&|or)\s*\(?[a-d1-4]\)?(?:\s|$)"
    r"|option\s*\(?[a-d1-4])", re.I)
OPT_RE = re.compile(r"[\(\[]([A-D])[\)\]]")
PREFIXES = [
    "Read the question carefully and answer.",
    "Answer the following exam question.",
    "Consider the following problem from a competitive examination.",
]

def positional(opts):
    return any(POSITIONAL.search(str(o)) for o in opts)

def derangement(n=4, tries=20):
    for _ in range(tries):
        p = list(range(n)); rng.shuffle(p)
        if p != list(range(n)):
            return p
    return list(range(n))[::-1]

def remap(ans, perm):
    """perm[new_idx] = old_idx ; gold letters -> new letters"""
    old_to_new = {old: new for new, old in enumerate(perm)}
    letters = re.findall(r"[A-D]", str(ans).upper())
    return "".join(sorted("ABCD"[old_to_new["ABCD".index(l)]] for l in letters))

def split_embedded(q):
    """JEE-Adv style '... (A) x (B) y (C) z (D) w' -> (stem, [4 options]) or None"""
    ms = list(OPT_RE.finditer(q))
    if [m.group(1) for m in ms] != list("ABCD"):
        return None
    stem = q[:ms[0].start()].rstrip()
    opts = [q[ms[i].end(): ms[i + 1].start() if i < 3 else len(q)].strip() for i in range(4)]
    br = ("[", "]") if ms[0].group(0)[0] == "[" else ("(", ")")
    return stem, opts, br

def shuffle_row(r):
    """returns dict(question, options, answer) or None if not applicable"""
    qtype = str(r["qtype"])
    opts = r["options"]
    if isinstance(opts, list) and len(opts) == 4:                  # NEET-PG, JEE-Main MCQ
        if positional(opts) or not re.fullmatch(r"[A-D]", str(r["answer"])):
            return None
        perm = derangement()
        return dict(question=r["question"], options=[opts[i] for i in perm],
                    answer=remap(r["answer"], perm), perm=perm)
    if r["exam"] == "JEE-Adv" and qtype in ("MCQ", "MCQ(multiple)"):  # options embedded in text
        sp = split_embedded(r["question"])
        if sp is None:
            return None
        stem, o, br = sp
        if positional(o) or not re.search(r"[A-D]", str(r["answer"])):
            return None
        perm = derangement()
        new = [o[i] for i in perm]
        q = stem + "\n\n" + "\n\n".join(f"{br[0]}{'ABCD'[i]}{br[1]} {new[i]}" for i in range(4))
        return dict(question=q, options=None, answer=remap(r["answer"], perm), perm=perm)
    return None

def make_twins(base):
    out = []
    for _, r in base.iterrows():
        meta = dict(orig_id=r["id"], exam=r["exam"], subject=r["subject"], qtype=r["qtype"])
        s = shuffle_row(r)
        if s:
            out.append({**meta, "perturbation": "option_shuffle", **s})
        out.append({**meta, "perturbation": "prefix_noise",
                    "question": rng.choice(PREFIXES) + "\n\n" + r["question"],
                    "options": r["options"], "answer": r["answer"], "perm": None})
    return pd.DataFrame(out)

if __name__ == "__main__":
    base = pd.read_json("benchmark_base.jsonl", lines=True)
    twins = make_twins(base)
    twins.to_json("perturbed_twins.jsonl", orient="records", lines=True, force_ascii=False)
    print("saved perturbed_twins.jsonl", len(twins))
    cov = twins.groupby(["exam", "perturbation"]).size().unstack(fill_value=0)
    cov["base_n"] = base.groupby("exam").size()
    print(cov)
```

```python
"""
Step 3: evaluate small open LLMs on originals vs perturbed twins.
Colab:  pip install -q transformers accelerate pandas
Needs : benchmark_base.jsonl, perturbed_twins.jsonl (same folder)
Output: results_<model>.csv + printed metrics

Scope (first pass): single-answer MCQs only (NEET-PG, JEE-Main single_correct, JEE-Adv MCQ).
Scoring = next-token logits over A/B/C/D (no generation, no answer-parsing errors).
Numerical / multi-correct rows need generation (CoT) -> separate pass.
"""
import re, sys
import pandas as pd
from scipy.stats import binomtest

MODELS = [  # ungated; verify repo ids once. Add/remove as GPU allows.
    "Qwen/Qwen2.5-1.5B-Instruct",
    "Qwen/Qwen2.5-3B-Instruct",
    "microsoft/Phi-3.5-mini-instruct",
    "HuggingFaceTB/SmolLM2-1.7B-Instruct",
]
LETTERS = "ABCD"
INSTR = "\n\nAnswer with a single letter (A, B, C or D) only."

def is_single_mcq(r):
    if not re.fullmatch(r"[A-D]", str(r["answer"])):
        return False
    return (r["exam"] == "NEET-PG" or
            (r["exam"] == "JEE-Main" and r["qtype"] == "single_correct") or
            (r["exam"] == "JEE-Adv" and r["qtype"] == "MCQ"))

def render(r):
    q = str(r["question"])
    opts = r["options"]
    if isinstance(opts, list) and len(opts) == 4:       # options stored separately
        q += "\n\n" + "\n".join(f"{LETTERS[i]}. {opts[i]}" for i in range(4))
    return q + INSTR                                    # else: options already inside question text

def load_eval_rows():
    base = pd.read_json("benchmark_base.jsonl", lines=True)
    assert base["id"].is_unique, "duplicate ids -> re-run compile_benchmark.py (fixed version)"
    base = base[base.apply(is_single_mcq, axis=1)].copy()
    base["orig_id"] = base["id"]; base["perturbation"] = "original"
    tw = pd.read_json("perturbed_twins.jsonl", lines=True)
    tw = pd.concat([tw, pd.read_json("structural_twins.jsonl", lines=True)], ignore_index=True)  # <- NEW
    tw = pd.concat([tw, pd.read_json("paraphrase_twins.jsonl", lines=True)], ignore_index=True)
    tw = tw[tw["orig_id"].isin(base["id"])]
    base["perm"] = None
    cols = ["orig_id", "exam", "subject", "qtype", "perturbation", "question", "options", "answer", "perm"]
    return pd.concat([base[cols], tw[cols]], ignore_index=True)

def run_model(name, rows):
    import torch
    from transformers import AutoTokenizer, AutoModelForCausalLM
    tok = AutoTokenizer.from_pretrained(name)
    model = AutoModelForCausalLM.from_pretrained(
        name, torch_dtype=torch.float16, device_map="auto").eval()
    ids = [[tok.encode(l, add_special_tokens=False)[0], tok.encode(" " + l, add_special_tokens=False)[-1]]
           for l in LETTERS]
    preds = []
    for _, r in rows.iterrows():
        msgs = [{"role": "user", "content": render(r)}]
        text = tok.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
        enc = tok(text, return_tensors="pt", truncation=True, max_length=3000).to(model.device)
        with torch.no_grad():
            logits = model(**enc, use_cache=False).logits[0, -1].float()
        scores = [max(logits[a].item(), logits[b].item()) for a, b in ids]
        preds.append(LETTERS[max(range(4), key=lambda i: scores[i])])
    del model; torch.cuda.empty_cache()
    out = rows.copy(); out["pred"] = preds
    # prediction expressed in ORIGINAL option order (so shuffled twins are comparable)
    out["pred_orig_space"] = [
        LETTERS[int(p[LETTERS.index(pr)])] if isinstance(p, list) else pr
        for pr, p in zip(preds, out["perm"])]
    out["correct"] = out["pred"] == out["answer"]
    return out

def metrics(df):
    o = df[df.perturbation == "original"].set_index("orig_id")
    res = []
    for p in sorted(set(df.perturbation) - {"original"}):
        t = df[df.perturbation == p].set_index("orig_id")
        j = o.join(t, lsuffix="_o", rsuffix="_t", how="inner")
        b, c = int((j.correct_o & ~j.correct_t).sum()), int((~j.correct_o & j.correct_t).sum())
        pval = binomtest(b, b + c, 0.5).pvalue if b + c > 0 else 1.0   # exact McNemar
        res.append(dict(perturbation=p, n=len(j),
                        acc_orig=j.correct_o.mean(), acc_twin=j.correct_t.mean(),
                        drop=j.correct_o.mean() - j.correct_t.mean(),
                        broke=(j.correct_o & ~j.correct_t).mean(),      # right -> wrong
                        fixed=(~j.correct_o & j.correct_t).mean(),      # wrong -> right
                        flip=(j.correct_o != j.correct_t).mean(),
                        same_choice=(j.pred_orig_space_o == j.pred_orig_space_t).mean(),  # chose same option?
                        p_mcnemar=pval))
    return pd.DataFrame(res).round(3)

if __name__ == "__main__":
    rows = load_eval_rows()
    print(rows.groupby(["exam", "perturbation"]).size().unstack(fill_value=0))
    for m in MODELS:
        try:
            out = run_model(m, rows)
        except Exception as e:
            print(f"[skip] {m}: {e}"); continue
        out.to_csv(f"results_{m.split('/')[-1]}.csv", index=False)
        print("\n==", m)
        print("overall original acc:", round(out[out.perturbation == 'original'].correct.mean(), 3))
        print("pred letter dist (original):", out[out.perturbation == 'original'].pred.value_counts(normalize=True).round(2).to_dict())
        for exam, g in out.groupby("exam"):
            print(f"-- {exam}"); print(metrics(g).to_string(index=False))
```

```python
"""
Step 4: paraphrase perturbation (wording changes, meaning/numbers/options fixed).
Colab: self-contained, just paste in a cell and run (needs only benchmark_base.jsonl in the working folder)
Output: paraphrase_twins.jsonl  +  paraphrase_audit.csv (40 pairs for MANUAL meaning check)

Only the question STEM is rewritten; options are untouched -> gold answer stays valid.
A paraphrase is kept only if it passes automatic checks (numbers, LaTeX, key terms, length, not-a-copy).
Automatic checks do NOT prove meaning is preserved -> audit paraphrase_audit.csv by hand and report it.
"""
import re, random
from collections import Counter
import pandas as pd

PARAPHRASER = "Qwen/Qwen2.5-3B-Instruct"   # NOTE: also an evaluated model -> mention self-preference caveat
BATCH = 8
PROMPT = ("Rewrite the following exam question so that the wording and sentence structure are different, "
          "but the meaning is exactly the same. Keep every number, unit, symbol, formula (including anything "
          "between $ signs) and technical term unchanged. Do not answer the question, do not add or remove "
          "information. Output only the rewritten question.\n\nQuestion:\n{stem}")

OPT_RE = re.compile(r"[\(\[]([A-D])[\)\]]")

def split_embedded(q):
    """'... [A] x [B] y [C] z [D] w' (or with parentheses) -> (stem, [4 options], bracket style) or None"""
    ms = list(OPT_RE.finditer(q))
    if [m.group(1) for m in ms] != list("ABCD"):
        return None
    stem = q[:ms[0].start()].rstrip()
    opts = [q[ms[i].end(): ms[i + 1].start() if i < 3 else len(q)].strip() for i in range(4)]
    br = ("[", "]") if ms[0].group(0)[0] == "[" else ("(", ")")
    return stem, opts, br

def is_single_mcq(r):
    if not re.fullmatch(r"[A-D]", str(r["answer"])):
        return False
    return (r["exam"] == "NEET-PG" or
            (r["exam"] == "JEE-Main" and r["qtype"] == "single_correct") or
            (r["exam"] == "JEE-Adv" and r["qtype"] == "MCQ"))

NUM = re.compile(r"\d+(?:\.\d+)?")
LATEX = re.compile(r"\$[^$]+\$")
WORD = re.compile(r"[A-Za-z]{7,}")
GENERIC = {"following", "statement", "statements", "correct", "incorrect", "regarding", "consider",
           "question", "respectively", "between", "possible", "options", "choose", "determine",
           "calculate", "assuming", "according", "denotes", "present", "obtained"}   # exam boilerplate, free to reword

def validate(orig, new):
    """returns 'ok' or the reason for rejection"""
    if not new or new.strip() == orig.strip():
        return "identical"
    if re.match(r"(?i)\s*(here is|here's|rewritten|sure|paraphras|question:)", new):
        return "meta"
    if Counter(NUM.findall(orig)) != Counter(NUM.findall(new)):
        return "numbers"
    if any(m not in new for m in LATEX.findall(orig)):
        return "latex"
    if not 0.5 <= len(new) / max(1, len(orig)) <= 1.8:
        return "length"
    ow = {w.lower() for w in WORD.findall(orig)} - GENERIC
    nw = {w.lower() for w in WORD.findall(new)}
    if ow and len(ow & nw) / len(ow) < 0.7:
        return "terms"
    a, b = set(orig.lower().split()), set(new.lower().split())
    if len(a & b) / max(1, len(a | b)) > 0.9:
        return "too_similar"
    return "ok"

def get_stem(r):
    """-> (stem, rebuild_fn) or None"""
    opts = r["options"]
    if isinstance(opts, list) and len(opts) == 4:
        return str(r["question"]), lambda s: s
    sp = split_embedded(str(r["question"]))
    if sp is None:
        return None
    stem, o, br = sp
    return stem, lambda s: s + "\n\n" + "\n\n".join(f"{br[0]}{'ABCD'[i]}{br[1]} {o[i]}" for i in range(4))

def generate_paraphrases(stems):
    import torch
    from transformers import AutoTokenizer, AutoModelForCausalLM
    tok = AutoTokenizer.from_pretrained(PARAPHRASER); tok.padding_side = "left"
    if tok.pad_token is None:
        tok.pad_token = tok.eos_token
    model = AutoModelForCausalLM.from_pretrained(PARAPHRASER, torch_dtype=torch.float16, device_map="auto").eval()
    order = sorted(range(len(stems)), key=lambda i: len(stems[i]))     # similar lengths per batch
    out = [None] * len(stems)
    for s in range(0, len(order), BATCH):
        idx = order[s:s + BATCH]
        texts = [tok.apply_chat_template([{"role": "user", "content": PROMPT.format(stem=stems[i])}],
                                         tokenize=False, add_generation_prompt=True) for i in idx]
        enc = tok(texts, return_tensors="pt", padding=True).to(model.device)
        with torch.no_grad():
            gen = model.generate(**enc, max_new_tokens=512, do_sample=False)
        for i, g in zip(idx, gen):
            out[i] = tok.decode(g[enc["input_ids"].shape[1]:], skip_special_tokens=True).strip()
        print(f"paraphrased {min(s + BATCH, len(order))}/{len(order)}", end="\r")
    del model; torch.cuda.empty_cache()
    return out

def build(base, paraphrase_fn):
    cand = []
    for _, r in base.iterrows():
        g = get_stem(r)
        if g and len(g[0]) >= 40:                     # very short stems: paraphrase is trivial/risky
            cand.append((r, g[0], g[1]))
    new = paraphrase_fn([c[1] for c in cand])
    rows, audit, stats = [], [], []
    for (r, stem, rebuild), p in zip(cand, new):
        why = validate(stem, p)
        stats.append((r["exam"], why))
        if why == "ok":
            rows.append(dict(orig_id=r["id"], exam=r["exam"], subject=r["subject"], qtype=r["qtype"],
                             perturbation="paraphrase", question=rebuild(p), options=r["options"],
                             answer=r["answer"], perm=None))
            audit.append(dict(orig_id=r["id"], exam=r["exam"], original=stem, paraphrase=p))
    return pd.DataFrame(rows), pd.DataFrame(audit), pd.DataFrame(stats, columns=["exam", "result"])

if __name__ == "__main__":
    base = pd.read_json("benchmark_base.jsonl", lines=True)
    base = base[base.apply(is_single_mcq, axis=1)]
    tw, audit, stats = build(base, generate_paraphrases)
    tw.to_json("paraphrase_twins.jsonl", orient="records", lines=True, force_ascii=False)
    audit.sample(min(40, len(audit)), random_state=42).to_csv("paraphrase_audit.csv", index=False)
    print("\nkept:", len(tw), "of", len(stats))
    print(stats.groupby(["exam", "result"]).size().unstack(fill_value=0))
    for _, a in audit.sample(min(3, len(audit)), random_state=1).iterrows():
        print("\nORIG:", a.original[:300], "\nNEW :", a.paraphrase[:300])
```

```python
"""
Step 5: STRUCTURAL perturbations (category C). Gold answer stays valid, fully deterministic, no LLM.
Colab: needs only benchmark_base.jsonl in the working folder.
Output: structural_twins.jsonl (same schema as perturbed_twins.jsonl) + structural_audit.csv

Perturbations (single-answer MCQs only):
  opts_first     : options block shown BEFORE the question stem
  opts_inline    : options squeezed into one line  "A) x   B) y   C) z   D) w"
  q_first        : last (question-like) sentence moved to the front of the stem
                   (skipped if it contains anaphora: this / above / it ... or stem has <2 sentences)
  template_wrap  : "Question: ... / Options: ..." exam-paper style wrapper

All variants put the options INSIDE `question` and set options=None, so the existing
render() in Step 3 (which only appends the 'single letter' instruction) works unchanged.
Option ORDER is never changed -> gold letter stays the same.
"""
import re
import pandas as pd

OPT_RE = re.compile(r"[\(\[]([A-D])[\)\]]")
LETTERS = "ABCD"


def split_embedded(q):
    ms = list(OPT_RE.finditer(q))
    if [m.group(1) for m in ms] != list(LETTERS):
        return None
    stem = q[:ms[0].start()].rstrip()
    opts = [q[ms[i].end(): ms[i + 1].start() if i < 3 else len(q)].strip() for i in range(4)]
    return stem, opts


def is_single_mcq(r):
    if not re.fullmatch(r"[A-D]", str(r["answer"])):
        return False
    return (r["exam"] == "NEET-PG" or
            (r["exam"] == "JEE-Main" and r["qtype"] == "single_correct") or
            (r["exam"] == "JEE-Adv" and r["qtype"] == "MCQ"))


def get_stem_opts(r):
    """-> (stem, [4 options]) or None"""
    opts = r["options"]
    if isinstance(opts, list) and len(opts) == 4:
        return str(r["question"]).strip(), [str(o).strip() for o in opts]
    return split_embedded(str(r["question"]))


# ---------- sentence handling (LaTeX-safe: never split inside $...$) ----------
SENT_SPLIT = re.compile(r"(?<=[.?!])\s+(?=[A-Z])")
QSTART = re.compile(r"^(find|what|which|calculate|determine|how|compute|evaluate|choose|identify)\b", re.I)
ANAPHORA = re.compile(r"\b(above|previous|this|these|those|such|it|its|they|their|then|thus|hence|same)\b", re.I)


def split_sentences(s):
    parts = SENT_SPLIT.split(s)
    merged = []
    for p in parts:                       # re-join pieces if a $ is left open
        if merged and merged[-1].count("$") % 2 == 1:
            merged[-1] += " " + p
        else:
            merged.append(p)
    return merged


def question_first(stem):
    sents = split_sentences(stem)
    if len(sents) < 2:
        return None
    last = sents[-1].strip()
    if not (QSTART.match(last) or last.endswith("?")):
        return None
    if ANAPHORA.search(last):
        return None
    new = last + " " + " ".join(s.strip() for s in sents[:-1])
    return new if new != stem else None


# ---------- layouts ----------
def lay_opts_first(stem, o):
    return "\n".join(f"{LETTERS[i]}. {o[i]}" for i in range(4)) + "\n\n" + stem


def lay_opts_inline(stem, o):
    return stem + "\n\n" + "   ".join(f"{LETTERS[i]}) {o[i]}" for i in range(4))


def lay_template(stem, o):
    return ("Question: " + stem + "\n\nOptions:\n" +
            "\n".join(f"({LETTERS[i]}) {o[i]}" for i in range(4)))


def lay_plain(stem, o):               # base layout used after stem rewrite
    return stem + "\n\n" + "\n".join(f"{LETTERS[i]}. {o[i]}" for i in range(4))


def make_structural(base):
    rows = []
    for _, r in base.iterrows():
        g = get_stem_opts(r)
        if g is None:
            continue
        stem, o = g
        meta = dict(orig_id=r["id"], exam=r["exam"], subject=r["subject"], qtype=r["qtype"],
                    options=None, answer=r["answer"], perm=None)
        variants = {
            "opts_first": lay_opts_first(stem, o),
            "opts_inline": lay_opts_inline(stem, o),
            "template_wrap": lay_template(stem, o),
        }
        qf = question_first(stem)
        if qf:
            variants["q_first"] = lay_plain(qf, o)
        for name, q in variants.items():
            rows.append({**meta, "perturbation": name, "question": q})
    return pd.DataFrame(rows)


if __name__ == "__main__":
    base = pd.read_json("benchmark_base.jsonl", lines=True)
    base = base[base.apply(is_single_mcq, axis=1)]
    tw = make_structural(base)
    tw.to_json("structural_twins.jsonl", orient="records", lines=True, force_ascii=False)
    print("saved structural_twins.jsonl", len(tw))
    print(tw.groupby(["exam", "perturbation"]).size().unstack(fill_value=0))
    # small manual audit of the only non-trivial one (q_first)
    qf = tw[tw.perturbation == "q_first"]
    if len(qf):
        qf.sample(min(30, len(qf)), random_state=42)[["orig_id", "question"]].to_csv(
            "structural_audit.csv", index=False)
        print("audit sample written:", min(30, len(qf)))
```

```python
"""
Step 6: evaluate a MODEL LADDER (+ optional Claude / Gemini via API) on originals vs ALL twins.
Colab:  pip install -q transformers accelerate bitsandbytes pandas scipy anthropic google-genai
Needs : benchmark_base.jsonl, perturbed_twins.jsonl, structural_twins.jsonl  (paraphrase_twins.jsonl optional)
Output: ladder_<model>.csv per model  (resumable: finished models are skipped)

Local models : next-token logits over A/B/C/D  (same scoring as Step 3)
API models   : generation, first standalone A-D letter is parsed (APIs give no usable logits)
               -> scoring method differs from local. See CALIBRATION note at bottom.

API keys: set env vars ANTHROPIC_API_KEY / GEMINI_API_KEY  (in Colab: os.environ[...] = userdata.get(...))
"""
import os, re, json, time
import pandas as pd
from concurrent.futures import ThreadPoolExecutor

# ----------------------------- CONFIG -----------------------------
RUN_LOCAL = True
RUN_API = False                     # flip to True once keys are set

# (hf_repo, size_in_B, load_4bit).  Same-family ladder keeps the scale comparison clean.
# 3B appears twice on purpose: fp16 vs 4-bit measures the quantisation effect itself,
# so you can tell whether 7B/14B (4-bit) differences come from size or from quantisation.
LOCAL = [
    ("Qwen/Qwen2.5-0.5B-Instruct", 0.5, False),
    ("Qwen/Qwen2.5-1.5B-Instruct", 1.5, False),
    ("Qwen/Qwen2.5-3B-Instruct",   3.0, False),
    ("Qwen/Qwen2.5-3B-Instruct",   3.0, True),     # control
    ("Qwen/Qwen2.5-7B-Instruct",   7.0, True),
    # ("Qwen/Qwen2.5-14B-Instruct", 14.0, True),    # needs ~10-12 GB VRAM in 4-bit (L4/A100 ok, T4 tight)
]

# (provider, model_id).  Verify the ids against the provider docs before running.
API = [
    ("anthropic", "claude-haiku-5-5"),
    ("anthropic", "claude-sonnet-5-5"),
    ("anthropic", "claude-opus-5-5"),
    # ("google", "<gemini-flash-model-id>"),
    # ("google", "<gemini-pro-model-id>"),
]
API_WORKERS = 8
MAX_LEN = 3000
# ------------------------------------------------------------------

LETTERS = "ABCD"
INSTR = "\n\nAnswer with a single letter (A, B, C or D) only."


def is_single_mcq(r):
    if not re.fullmatch(r"[A-D]", str(r["answer"])):
        return False
    return (r["exam"] == "NEET-PG" or
            (r["exam"] == "JEE-Main" and r["qtype"] == "single_correct") or
            (r["exam"] == "JEE-Adv" and r["qtype"] == "MCQ"))


def render(r):
    q = str(r["question"])
    opts = r["options"]
    if isinstance(opts, list) and len(opts) == 4:
        q += "\n\n" + "\n".join(f"{LETTERS[i]}. {opts[i]}" for i in range(4))
    return q + INSTR


def load_eval_rows():
    base = pd.read_json("benchmark_base.jsonl", lines=True)
    assert base["id"].is_unique
    base = base[base.apply(is_single_mcq, axis=1)].copy()
    base["orig_id"] = base["id"]; base["perturbation"] = "original"; base["perm"] = None
    files = ["perturbed_twins.jsonl", "structural_twins.jsonl", "paraphrase_twins.jsonl"]
    tw = pd.concat([pd.read_json(f, lines=True) for f in files if os.path.exists(f)], ignore_index=True)
    tw = tw[tw["orig_id"].isin(base["id"])]
    cols = ["orig_id", "exam", "subject", "qtype", "perturbation", "question", "options", "answer", "perm"]
    return pd.concat([base[cols], tw[cols]], ignore_index=True)


def finish(rows, preds, tag, size, kind):
    out = rows.copy()
    out["pred"] = preds
    # prediction expressed in ORIGINAL option order so shuffled twins are comparable
    out["pred_orig_space"] = [
        LETTERS[int(p[LETTERS.index(pr)])] if (isinstance(p, list) and pr in LETTERS) else pr
        for pr, p in zip(preds, out["perm"])]
    out["correct"] = out["pred"] == out["answer"]
    out["model"], out["size_b"], out["kind"] = tag, size, kind
    return out


# ----------------------------- local -----------------------------
def run_local(name, size, quant, rows):
    import torch
    from transformers import AutoTokenizer, AutoModelForCausalLM
    tok = AutoTokenizer.from_pretrained(name)
    kw = dict(device_map="auto")
    if quant:
        from transformers import BitsAndBytesConfig
        kw["quantization_config"] = BitsAndBytesConfig(
            load_in_4bit=True, bnb_4bit_quant_type="nf4", bnb_4bit_compute_dtype=torch.float16)
    else:
        kw["torch_dtype"] = torch.float16
    model = AutoModelForCausalLM.from_pretrained(name, **kw).eval()
    ids = [[tok.encode(l, add_special_tokens=False)[0], tok.encode(" " + l, add_special_tokens=False)[-1]]
           for l in LETTERS]
    preds = []
    for n, (_, r) in enumerate(rows.iterrows()):
        text = tok.apply_chat_template([{"role": "user", "content": render(r)}],
                                       tokenize=False, add_generation_prompt=True)
        enc = tok(text, return_tensors="pt", truncation=True, max_length=MAX_LEN).to(model.device)
        with torch.no_grad():
            try:
                lg = model(**enc, use_cache=False, logits_to_keep=1).logits[0, -1].float()
            except TypeError:                      # older transformers
                lg = model(**enc, use_cache=False).logits[0, -1].float()
        sc = [max(lg[a].item(), lg[b].item()) for a, b in ids]
        preds.append(LETTERS[max(range(4), key=lambda i: sc[i])])
        if n % 500 == 0:
            print(f"  {n}/{len(rows)}", end="\r")
    del model; torch.cuda.empty_cache()
    return preds


# ------------------------------ API ------------------------------
_LETTER = re.compile(r"\b([ABCD])\b")


def parse_letter(text):
    m = _LETTER.search((text or "").upper())
    return m.group(1) if m else "X"            # X = unparseable -> counted wrong


def ask(provider, model, prompt, state):
    for attempt in range(6):
        try:
            if provider == "anthropic":
                import anthropic
                state.setdefault("c", anthropic.Anthropic())
                kw = dict(model=model, max_tokens=16, messages=[{"role": "user", "content": prompt}])
                if state.get("temp", True):
                    kw["temperature"] = 0
                r = state["c"].messages.create(**kw)
                return "".join(b.text for b in r.content if getattr(b, "type", "") == "text")
            else:
                from google import genai
                from google.genai import types
                state.setdefault("c", genai.Client(api_key=os.environ["GEMINI_API_KEY"]))
                cfg = dict(temperature=0, max_output_tokens=256)
                r = state["c"].models.generate_content(
                    model=model, contents=prompt, config=types.GenerateContentConfig(**cfg))
                return r.text or ""
        except Exception as e:
            if "temperature" in str(e).lower():
                state["temp"] = False              # model rejects temperature -> drop it
            time.sleep(min(2 ** attempt, 30))
    return ""


def run_api(provider, model, rows, tag):
    cache_f = f"cache_{tag}.json"
    cache = json.load(open(cache_f)) if os.path.exists(cache_f) else {}
    state = {}
    prompts = [render(r) for _, r in rows.iterrows()]
    todo = [i for i in range(len(rows)) if str(i) not in cache]
    for s in range(0, len(todo), 200):
        chunk = todo[s:s + 200]
        with ThreadPoolExecutor(API_WORKERS) as ex:
            res = list(ex.map(lambda i: ask(provider, model, prompts[i], state), chunk))
        for i, t in zip(chunk, res):
            cache[str(i)] = parse_letter(t)
        json.dump(cache, open(cache_f, "w"))
        print(f"  {min(s + 200, len(todo))}/{len(todo)}", end="\r")
    return [cache[str(i)] for i in range(len(rows))]


# ------------------------------ main ------------------------------
if __name__ == "__main__":
    rows = load_eval_rows()
    print(rows.groupby(["exam", "perturbation"]).size().unstack(fill_value=0))

    if RUN_LOCAL:
        for name, size, quant in LOCAL:
            tag = name.split("/")[-1] + ("@4bit" if quant else "")
            f = f"ladder_{tag}.csv"
            if os.path.exists(f):
                print("[skip, exists]", tag); continue
            print("\n==", tag)
            try:
                preds = run_local(name, size, quant, rows)
            except Exception as e:
                print(f"[fail] {tag}: {e}"); continue
            out = finish(rows, preds, tag, size, "local")
            out.to_csv(f, index=False)
            print("original acc:", round(out[out.perturbation == "original"].correct.mean(), 3))

    if RUN_API:
        for provider, model in API:
            f = f"ladder_{model}.csv"
            if os.path.exists(f):
                print("[skip, exists]", model); continue
            print("\n==", model)
            preds = run_api(provider, model, rows, model)
            out = finish(rows, preds, model, float("nan"), "api")
            out.to_csv(f, index=False)
            print("original acc:", round(out[out.perturbation == "original"].correct.mean(), 3),
                  "| unparseable:", round((out.pred == "X").mean(), 4))

# CALIBRATION: local = logit scoring, API = generation + parsing. To show the gap is not a
# scoring artefact, run ONE local model both ways (e.g. Qwen2.5-3B with model.generate +
# parse_letter) on the original rows and report that the accuracies match within ~1-2 points.
```

```python
"""
Step 7: analysis + figures from ladder_*.csv  (output of step6_ladder_eval.py)
Colab: just paste and run. Writes figs/*.png, summary.csv, trend_table.csv

Questions are the unit of resampling (cluster bootstrap), so pooling several perturbations of the
same question does not make the confidence intervals look tighter than they are.

Figures
 F1  flip rate vs model size, one panel per perturbation (with 95% CI)
 F2  pooled flip rate vs size + noise floor (prefix_noise) + excess over floor
 F3  original accuracy vs size per exam (chance line)
 F4  conditional consistency  P(correct on twin | correct on original) vs size
 F5  heatmap model x perturbation : flip rate
 F6  heatmap model x perturbation : same_choice
 F7  pooled flip vs size, split by exam
 F8  broke vs fixed per (model, perturbation)  -> direction of damage
 F9  predicted-letter distribution per model vs gold distribution (position bias)
 F10 excess flip over noise floor, per perturbation vs size
 F11 accuracy vs flip scatter (are bigger models both better and steadier?)
"""
import os, glob
import numpy as np, pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

OUT = "figs"; os.makedirs(OUT, exist_ok=True)
rng = np.random.default_rng(0)
B = 1000
API_ORDER = ["claude-haiku-5-5", "claude-sonnet-5-5", "claude-opus-5-5"]
NOISE = "prefix_noise"
plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": .3})


# ------------------------------ data ------------------------------
def load():
    fs = sorted(glob.glob("ladder_*.csv"))
    assert fs, "no ladder_*.csv found"
    return pd.concat([pd.read_csv(f) for f in fs], ignore_index=True)


def build_pairs(df):
    rows = []
    for m, g in df.groupby("model"):
        o = g[g.perturbation == "original"].drop_duplicates("orig_id").set_index("orig_id")
        for p, t in g[g.perturbation != "original"].groupby("perturbation"):
            t = t.drop_duplicates("orig_id").set_index("orig_id")
            j = o.join(t, lsuffix="_o", rsuffix="_t", how="inner")
            rows.append(pd.DataFrame({
                "model": m, "perturbation": p, "orig_id": j.index, "exam": j["exam_o"].values,
                "co": j["correct_o"].astype(bool).values, "ct": j["correct_t"].astype(bool).values,
                "same": (j["pred_orig_space_o"] == j["pred_orig_space_t"]).values}))
    P = pd.concat(rows, ignore_index=True)
    P["flip"] = P.co != P.ct
    P["broke"] = P.co & ~P.ct
    P["fixed"] = ~P.co & P.ct
    return P


def boot(x):
    x = np.asarray(x, float)
    idx = rng.integers(0, len(x), (B, len(x)))
    m = x[idx].mean(1)
    return x.mean(), *np.percentile(m, [2.5, 97.5])


def summarize(P, by, meta):
    out = []
    for k, g in P.groupby(by):
        k = k if isinstance(k, tuple) else (k,)
        q = g.groupby("orig_id").flip.mean()                   # per-question flip -> cluster bootstrap
        f, lo, hi = boot(q.values)
        out.append(dict(zip(by, k), n=g.orig_id.nunique(), flip=f, flip_lo=lo, flip_hi=hi,
                        same=g.same.mean(), acc_o=g.groupby("orig_id").co.first().mean(),
                        acc_t=g.ct.mean(), cond=g.ct[g.co].mean() if g.co.any() else np.nan,
                        broke=g.broke.mean(), fixed=g.fixed.mean()))
    S = pd.DataFrame(out)
    return S.merge(meta, on="model")


def make_xmap(meta):
    loc = meta[meta.kind == "local"].sort_values("size_b")
    xm = {}
    for _, r in loc.iterrows():
        xm[r.model] = r.size_b * (1.08 if r.model.endswith("@4bit") else 1.0)
    base = (loc.size_b.max() if len(loc) else 1) * 2.5
    api = [m for m in API_ORDER if m in set(meta.model)] + \
          sorted(set(meta[meta.kind == "api"].model) - set(API_ORDER))
    for k, m in enumerate(api):
        xm[m] = base * 1.6 ** k
    return xm


def style_axis(ax, xm):
    ax.set_xscale("log")
    ax.set_xticks(list(xm.values()))
    ax.set_xticklabels([m.replace("-Instruct", "").replace("Qwen2.5-", "Q") for m in xm],
                       rotation=60, ha="right", fontsize=7)
    ax.minorticks_off()


def line(ax, sub, y, xm, lo=None, hi=None, color="C0", label=None):
    if sub.empty:
        return
    sub = sub.assign(x=sub.model.map(xm)).sort_values("x")
    q4 = sub.model.str.endswith("@4bit")
    main = sub[(sub.kind == "local") & ~q4]
    ax.plot(main.x, main[y], "-o", color=color, label=label)
    ax.scatter(sub[q4].x, sub[q4][y], facecolors="none", edgecolors=color, s=60, zorder=3)
    ax.scatter(sub[sub.kind == "api"].x, sub[sub.kind == "api"][y], marker="s", color=color, zorder=3)
    if lo:
        ax.errorbar(sub.x, sub[y], yerr=[sub[y] - sub[lo], sub[hi] - sub[y]],
                    fmt="none", ecolor=color, alpha=.5, capsize=2)


def save(fig, name):
    fig.tight_layout(); fig.savefig(f"{OUT}/{name}.png"); plt.close(fig); print("saved", name)


def heat(M, title, name, fmt="{:.2f}", cmap="viridis"):
    fig, ax = plt.subplots(figsize=(1.0 * M.shape[1] + 3, 0.5 * M.shape[0] + 2))
    im = ax.imshow(M.values, aspect="auto", cmap=cmap)
    ax.set_xticks(range(M.shape[1])); ax.set_xticklabels(M.columns, rotation=40, ha="right")
    ax.set_yticks(range(M.shape[0])); ax.set_yticklabels(M.index)
    for i in range(M.shape[0]):
        for j in range(M.shape[1]):
            v = M.values[i, j]
            if not np.isnan(v):
                ax.text(j, i, fmt.format(v), ha="center", va="center", color="w", fontsize=8)
    ax.grid(False); ax.set_title(title); fig.colorbar(im, ax=ax)
    save(fig, name)


# ------------------------------ main ------------------------------
if __name__ == "__main__":
    df = load()
    meta = df.drop_duplicates("model")[["model", "size_b", "kind"]].reset_index(drop=True)
    xm = make_xmap(meta)
    order = sorted(meta.model, key=lambda m: xm[m])
    P = build_pairs(df)
    perts = sorted(P.perturbation.unique())

    S = summarize(P, ["model", "perturbation"], meta)
    A = summarize(P, ["model"], meta)                                  # pooled over all perturbations
    E = summarize(P, ["model", "exam"], meta)                          # pooled, per exam
    S.to_csv("summary.csv", index=False)
    nf = S[S.perturbation == NOISE].set_index("model").flip

    # F1
    n = len(perts); cols = 3; rws = int(np.ceil(n / cols))
    fig, axs = plt.subplots(rws, cols, figsize=(4.6 * cols, 3.4 * rws), squeeze=False)
    for ax, p in zip(axs.ravel(), perts):
        line(ax, S[S.perturbation == p], "flip", xm, "flip_lo", "flip_hi")
        ax.set_title(p, fontsize=10); ax.set_ylabel("flip rate"); style_axis(ax, xm)
    for ax in axs.ravel()[n:]:
        ax.axis("off")
    fig.suptitle("F1  Flip rate vs model size (95% CI; hollow = 4-bit, square = API)")
    save(fig, "F1_flip_vs_size_per_perturbation")

    # F2
    fig, axs = plt.subplots(1, 2, figsize=(11, 4))
    line(axs[0], A, "flip", xm, "flip_lo", "flip_hi", "C3", "all perturbations pooled")
    line(axs[0], S[S.perturbation == NOISE], "flip", xm, color="gray", label="noise floor (prefix_noise)")
    axs[0].set_ylabel("flip rate"); axs[0].legend(fontsize=8); style_axis(axs[0], xm)
    A2 = A.assign(excess=A.flip - A.model.map(nf))
    line(axs[1], A2, "excess", xm, color="C3"); axs[1].axhline(0, color="k", lw=.8)
    axs[1].set_ylabel("flip - noise floor"); style_axis(axs[1], xm)
    fig.suptitle("F2  Pooled flip rate and excess over noise floor")
    save(fig, "F2_pooled_flip_and_excess")

    # F3
    fig, ax = plt.subplots(figsize=(6.5, 4))
    for k, ex in enumerate(sorted(E.exam.unique())):
        line(ax, E[E.exam == ex], "acc_o", xm, color=f"C{k}", label=ex)
    ax.axhline(.25, color="k", ls="--", lw=.8, label="chance (0.25)")
    ax.set_ylabel("original accuracy"); ax.legend(fontsize=8); style_axis(ax, xm)
    ax.set_title("F3  Original accuracy vs size")
    save(fig, "F3_accuracy_vs_size")

    # F4
    fig, ax = plt.subplots(figsize=(7, 4.2))
    for k, p in enumerate(perts):
        line(ax, S[S.perturbation == p], "cond", xm, color=f"C{k % 10}", label=p)
    ax.set_ylabel("P(correct on twin | correct on original)"); ax.legend(fontsize=7, ncol=2); style_axis(ax, xm)
    ax.set_title("F4  Conditional consistency vs size")
    save(fig, "F4_conditional_consistency")

    # F5 / F6
    heat(S.pivot(index="model", columns="perturbation", values="flip").loc[order],
         "F5  Flip rate (lower = more robust)", "F5_heatmap_flip", cmap="magma")
    heat(S.pivot(index="model", columns="perturbation", values="same").loc[order],
         "F6  same_choice (higher = more consistent)", "F6_heatmap_same_choice", cmap="viridis")

    # F7
    fig, ax = plt.subplots(figsize=(6.5, 4))
    for k, ex in enumerate(sorted(E.exam.unique())):
        line(ax, E[E.exam == ex], "flip", xm, "flip_lo", "flip_hi", f"C{k}", ex)
    ax.set_ylabel("pooled flip rate"); ax.legend(fontsize=8); style_axis(ax, xm)
    ax.set_title("F7  Pooled flip rate by exam")
    save(fig, "F7_flip_by_exam")

    # F8
    fig, ax = plt.subplots(figsize=(5.5, 5))
    for k, m in enumerate(order):
        s = S[S.model == m]
        ax.scatter(s.broke, s.fixed, label=m, color=f"C{k % 10}", alpha=.8)
    mx = max(S.broke.max(), S.fixed.max()) * 1.05
    ax.plot([0, mx], [0, mx], "k--", lw=.8)
    ax.set_xlabel("broke (right -> wrong)"); ax.set_ylabel("fixed (wrong -> right)")
    ax.legend(fontsize=6); ax.set_title("F8  Direction of change (below diagonal = net damage)")
    save(fig, "F8_broke_vs_fixed")

    # F9
    orig = df[df.perturbation == "original"]
    L = list("ABCD")
    D = {m: orig[orig.model == m].pred.value_counts(normalize=True).reindex(L + ["X"]).fillna(0) for m in order}
    gold = orig.drop_duplicates("orig_id").answer.value_counts(normalize=True).reindex(L).fillna(0)
    fig, ax = plt.subplots(figsize=(1.0 * len(order) + 3, 4))
    bottom = np.zeros(len(order))
    for k, l in enumerate(L + ["X"]):
        v = np.array([D[m][l] for m in order])
        ax.bar(range(len(order)), v, bottom=bottom, label=l if l != "X" else "unparseable"); bottom += v
    ax.set_xticks(range(len(order))); ax.set_xticklabels(order, rotation=60, ha="right", fontsize=7)
    ax.set_ylabel("share of predictions"); ax.legend(fontsize=7)
    ax.set_title("F9  Predicted letters (original). Gold: " + " ".join(f"{l}={gold[l]:.2f}" for l in L))
    save(fig, "F9_letter_distribution")

    # F10
    Sx = S[S.perturbation != NOISE].copy()
    Sx["excess"] = Sx.flip - Sx.model.map(nf)
    fig, ax = plt.subplots(figsize=(7, 4.2))
    for k, p in enumerate(sorted(Sx.perturbation.unique())):
        line(ax, Sx[Sx.perturbation == p], "excess", xm, color=f"C{k % 10}", label=p)
    ax.axhline(0, color="k", lw=.8); ax.set_ylabel("flip - noise floor")
    ax.legend(fontsize=7, ncol=2); style_axis(ax, xm); ax.set_title("F10  Excess flip over noise floor")
    save(fig, "F10_excess_flip")

    # F11
    fig, ax = plt.subplots(figsize=(5.8, 4.5))
    for k, (_, r) in enumerate(A.iterrows()):
        ax.scatter(r.flip, r.acc_o, s=60, color=f"C{k % 10}", marker="s" if r.kind == "api" else "o")
        ax.annotate(r.model.replace("-Instruct", ""), (r.flip, r.acc_o), fontsize=6, xytext=(3, 3),
                    textcoords="offset points")
    ax.set_xlabel("pooled flip rate (lower = steadier)"); ax.set_ylabel("original accuracy")
    ax.set_title("F11  Better AND steadier? (top-left is best)")
    save(fig, "F11_accuracy_vs_flip")

    # ---------------- trend table: Qwen fp16 ladder only ----------------
    ladder = meta[(meta.kind == "local") & ~meta.model.str.endswith("@4bit")]
    rows = []
    for p, g in [("ALL", A)] + [(p, S[S.perturbation == p]) for p in perts]:
        g = g[g.model.isin(ladder.model)].sort_values("size_b")
        if len(g) >= 3:
            rho, pv = spearmanr(np.log(g.size_b), g.flip)
            pv = pv if len(g) >= 5 else np.nan          # p-value meaningless with <5 points
            rows.append(dict(perturbation=p, n_models=len(g), spearman_rho=rho, p=pv,
                             flip_smallest=g.flip.iloc[0], flip_largest=g.flip.iloc[-1]))
    T = pd.DataFrame(rows).round(3)
    T.to_csv("trend_table.csv", index=False)
    print("\nTREND (flip vs log size; rho<0 supports 'bigger = steadier'; few models -> descriptive only)")
    print(T.to_string(index=False))
    chance = A[A.acc_o < 0.30].model.tolist()
    if chance:
        print("\n[flag] original accuracy < 0.30 (near chance, flips are mostly noise):", chance)
```
